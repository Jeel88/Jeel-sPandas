```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"],
    "Subject": ["Python", "Python", "Java", "Python", "Java"]
})

print("Original DataFrame:")
print(df)
```

```python
df["Marks_Double"] = df["Marks"].apply(lambda x: x * 2)

print("\nMarks Doubled:")
print(df)
```

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"


df["Grade"] = df["Marks"].apply(grade)

print("\nGrade:")
print(df[["Name", "Marks", "Grade"]])
```

```python
city_code = {
    "Mumbai": "MUM",
    "Pune": "PUN",
    "Surat": "SUR",
    "Delhi": "DEL"
}

df["City_Code"] = df["City"].map(city_code)

print("\nCity Codes:")
print(df[["City", "City_Code"]])
```

```python
df["Subject"] = df["Subject"].replace({
    "Python": "Python Programming",
    "Java": "Java Programming"
})

print("\nAfter Replace:")
print(df[["Name", "Subject"]])
```

```python
df["Rank"] = df["Marks"].rank(
    ascending=False
)

print("\nStudent Rank:")
print(df[["Name", "Marks", "Rank"]])
```

```python
duplicate_df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Jeel", "Amit"],
    "Marks": [85, 92, 85, 76]
})

print("\nDuplicate Rows:")
print(duplicate_df.duplicated())

print("\nRemove Duplicates:")
print(duplicate_df.drop_duplicates())
```

```python
print("\nUnique Cities:")
print(df["City"].unique())

print("\nNumber of Unique Cities:")
print(df["City"].nunique())
```

```python
df_copy = df.copy()

df_copy["Marks"] = df_copy["Marks"] + 5

print("\nOriginal Marks:")
print(df["Marks"])

print("\nCopied DataFrame Marks:")
print(df_copy["Marks"])
```

```python
df["Result"] = np.where(
    df["Marks"] >= 40,
    "Pass",
    "Fail"
)

print("\nResult:")
print(df[["Name", "Marks", "Result"]])
```

```python
conditions = [
    df["Marks"] >= 90,
    df["Marks"] >= 80,
    df["Marks"] >= 70
]

choices = [
    "Excellent",
    "Very Good",
    "Good"
]

df["Performance"] = np.select(
    conditions,
    choices,
    default="Needs Improvement"
)

print("\nPerformance:")
print(df[["Name", "Marks", "Performance"]])
```

```python
print("\nTop 3 Students:")
print(df.nlargest(3, "Marks")[["Name", "Marks"]])
```

```python
result = df.query("Marks > 80 and Age >= 20")

print("\nMarks > 80 and Age >= 20:")
print(result[["Name", "Age", "Marks"]])
```

```python
grouped = df.groupby("City")["Marks"].agg(
    ["mean", "min", "max", "count"]
)

print("\nCity-wise Analysis:")
print(grouped)
```

```python
df["City_Average"] = df.groupby("City")["Marks"].transform("mean")

print("\nCity Average for Each Student:")
print(df[["Name", "City", "Marks", "City_Average"]])
```

```python
df["Difference"] = (
    df["Marks"] - df["City_Average"]
)

print("\nDifference from City Average:")
print(df[[
    "Name",
    "City",
    "Marks",
    "City_Average",
    "Difference"
]])
```

```python
df = df.reset_index(drop=True)

print("\nReset Index:")
print(df)
```

```python
print("\nFinal DataFrame:")
print(df)
```
