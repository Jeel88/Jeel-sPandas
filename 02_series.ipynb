```python
import pandas as pd
```

```python
marks = pd.Series([10,20,30,40,50,60])
print(marks)
```

```python
marks = pd.Series(
    [85, 92, 76, 88],
    index=["Jeel", "Heet", "Daksh", "Darsh"]
)
print(marks)
```

```python
print(marks["Heet"])
```

```python
marks = pd.Series([85, 92, 76, 88])
print(marks[2])
```

```python
marks = {
    "Jeel": 85,
    "Heet": 92,
    "Daksh": 76,
    "Darsh": 88
}
series = pd.Series(marks)
print(series)
```

```python
print(series.dtype)
```

```python
marks = pd.Series([85, 92, 76, 88])
print(marks +5)
```

```python
marks.mean()
marks.max()
marks.min()
marks.sum()
marks.count()
marks.std()
print(marks.mean())
```

```python
marks = pd.Series([85, 92, 76, 88])
print(marks>80)
print(marks[marks>80])
```

```python
print(marks.iloc[2])
```

```python
print(marks.iloc[1:4])
```
