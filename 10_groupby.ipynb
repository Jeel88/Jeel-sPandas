```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha", "Karan"],
    "Age": [20, 21, 19, 20, 22, 21],
    "Marks": [85, 92, 76, 88, 95, 82],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi", "Pune"]
})

df
```

```python
df.groupby("City")
```

```python
df.groupby("City")["Marks"].mean()
```

```python
df.groupby("City")["Marks"].sum()
```

```python
df.groupby("City")["Name"].count()
```

```python
df["City"].value_counts()
```

```python
df.groupby("City")["Name"].count()
```

```python
df.groupby(["City", "Age"])["Marks"].mean()
```

```python
df.groupby("City")["Marks"].mean().sort_values(ascending=False)
```

```python
result = df.groupby("City")["Marks"].mean().reset_index()
```

| Task                             | Code                                                              |
| -------------------------------- | ----------------------------------------------------------------- |
| Group by column                  | `df.groupby("City")`                                              |
| Average                          | `df.groupby("City")["Marks"].mean()`                              |
| Sum                              | `df.groupby("City")["Marks"].sum()`                               |
| Minimum                          | `df.groupby("City")["Marks"].min()`                               |
| Maximum                          | `df.groupby("City")["Marks"].max()`                               |
| Count                            | `df.groupby("City")["Name"].count()`                              |
| Multiple statistics              | `df.groupby("City")["Marks"].agg(["mean","min","max"])`           |
| Multiple grouping columns        | `df.groupby(["City","Age"])["Marks"].mean()`                      |
| Multiple columns                 | `df.groupby("City")[["Marks","Age"]].mean()`                      |
| Sort grouped result              | `df.groupby("City")["Marks"].mean().sort_values(ascending=False)` |
| Convert result back to DataFrame | `.reset_index()`                                                  |
