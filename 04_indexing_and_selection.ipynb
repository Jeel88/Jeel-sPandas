```python
import pandas as pd
```

```python
data = {
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.loc[0])
```

```python
print(df.iloc[0])
```

```python
df.loc[0,"Marks"]
```

```python
df.iloc[0,2]
```

```python
df.loc[:, ["Name", "Marks"]]
```

```python
df.loc[[0,2,4]]
```

```python
df.iloc[[0,2,4],[0,3]]
```

```python
df.iloc[1:4]
```

```python
df.loc[1:4]
```

```python
df.iloc[:, 0:3]
```

```python
df.index = ["A", "B", "C", "D", "E"]
```

```python
df.loc["B", "Marks"]
```

```python
df.loc["C"]
```

```python
df.iloc[2]
```

**Cheat Sheet**<br>
.loc<br>
│<br>
├── label based<br>
├── df.loc["A"]<br>
├── df.loc["A", "Marks"]<br>
├── df.loc[:, "Marks"]<br>
├── df.loc[:, ["Name", "Marks"]]<br>
└── df.loc["A":"C"]<br>
<br><br>
.iloc<br>
│<br>
├── position based<br>
├── df.iloc[0]<br>
├── df.iloc[0, 2]<br>
├── df.iloc[:, 2]<br>
├── df.iloc[:, [0, 2]]<br>
└── df.iloc[0:3]<br>
