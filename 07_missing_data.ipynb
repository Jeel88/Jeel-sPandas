```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, np.nan, 20, 22],
    "Marks": [85, np.nan, 76, 88, np.nan],
    "City": ["Mumbai", "Pune", "Surat", np.nan, "Delhi"]
})

df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df[df.isna().any(axis=1)]
```

```python
df[df.notna().all(axis=1)]
```

```python
df.dropna(subset=["Marks"])
```

```python
df.dropna(axis=1)
```

```python
df["Marks"] = df["Marks"].fillna(0)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].median())
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
df.isna().sum()
```

| Task                                 | Code                                       |
| ------------------------------------ | ------------------------------------------ |
| Find missing values                  | `df.isna()`                                |
| Count missing values                 | `df.isna().sum()`                          |
| Check non-missing values             | `df.notna()`                               |
| Find rows with any missing value     | `df[df.isna().any(axis=1)]`                |
| Find rows with no missing values     | `df[df.notna().all(axis=1)]`               |
| Remove rows with missing values      | `df.dropna()`                              |
| Remove columns with missing values   | `df.dropna(axis=1)`                        |
| Remove rows based on specific column | `df.dropna(subset=["Marks"])`              |
| Fill missing values                  | `df.fillna(0)`                             |
| Fill column with a value             | `df["Marks"].fillna(0)`                    |
| Fill with mean                       | `df["Age"].fillna(df["Age"].mean())`       |
| Fill with median                     | `df["Marks"].fillna(df["Marks"].median())` |
| Fill with mode                       | `df["City"].fillna(df["City"].mode()[0])`  |
