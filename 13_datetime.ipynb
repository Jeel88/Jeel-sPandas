```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya"],
    "Joining_Date": [
        "2025-06-15",
        "2025-07-20",
        "2026-01-10",
        "2026-03-25"
    ]
})

df
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

```python
df["Joining_Date"].dt.year
```

```python
df["Joining_Date"].dt.month
```

```python
df["Joining_Date"].dt.day
```

```python
df["Joining_Date"].dt.day_name()
```

```python
df["Year"] = df["Joining_Date"].dt.year
df["Month"] = df["Joining_Date"].dt.month
df["Day"] = df["Joining_Date"].dt.day
```

```python
df[df["Joining_Date"] > "2026-01-01"]
```

```python
df.sort_values("Joining_Date", ascending=False)
```
