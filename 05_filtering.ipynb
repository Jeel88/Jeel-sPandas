**Most Imp** : df[condition]


```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})
print(df)
```

```python
df["Marks"] > 80
```

```python
df[df["Marks"] > 80]
```

```python
df[df["City"] == "Mumbai"]
```

```python
df[(df["City"] == "Mumbai") | (df["City"] == "Pune")]
```

```python
df[(df["City"] == "Mumbai") | (df["City"] == "Pune")]
```

```python
df[df["City"].isin(["Mumbai", "Pune", "Delhi"])]
```

```python
df[df["Marks"].between(80, 90)]
```

```python
df[df["Name"].str.startswith("J")]
df[df["Name"].str.endswith("a")]
```

```python
df.loc[
    df["Marks"] > 80,
    ["Name", "Marks"]
]
```

```python
df[df["Marks"].notna()]
```

```python
df[df["Marks"].isna()]
```
