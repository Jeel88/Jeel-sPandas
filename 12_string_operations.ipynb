```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})

df
```

```python
df["Name"].str.lower()
```

```python
df["Name"].str.upper()
```

```python
df["Name"].str.len()
```

```python
df["City"].str.contains("Mumbai")
```

```python
df["Name"].str.startswith("J")
```

```python
df["Name"].str.endswith("h")
```

```python
df["City"] = df["City"].str.replace("Mumbai", "Bombay")
```

```python
df["Name"].str.strip()
```

```python
df
```

```python
df = pd.DataFrame({
    "Full_Name": ["Jeel Savaliya", "Rahul Patel", "Amit Shah"]
})
```

```python
df["Full_Name"].str.split(" ")
```

```python
df["Full_Name"].str.split(" ").str[0]
```

```python
df = pd.DataFrame({
    "Email": [
        "jeel@gmail.com",
        "rahul@yahoo.com",
        "amit@gmail.com"
    ]
})
```

```python
df["Email"].str.split("@").str[1]
```

```python
df["Email"].str.count("@")
```

| Operation     | Code                     |
| ------------- | ------------------------ |
| Lowercase     | `.str.lower()`           |
| Uppercase     | `.str.upper()`           |
| Length        | `.str.len()`             |
| Contains      | `.str.contains("x")`     |
| Starts with   | `.str.startswith("x")`   |
| Ends with     | `.str.endswith("x")`     |
| Replace       | `.str.replace("x", "y")` |
| Remove spaces | `.str.strip()`           |
| Split         | `.str.split()`           |
| Count         | `.str.count("x")`        |
