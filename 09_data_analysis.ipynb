```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})

df
```

```python
df["Marks"].mean()
```

```python
df["Marks"].median()
```

```python
df["Age"].mode()
```

```python
df["Marks"].sum()
```

```python
df["Marks"].max()
```

```python
df["Marks"].min()
```

```python
df["Marks"].std()
```

```python
df["Marks"].var()
```

```python
df["Marks"].describe()
```

```python
df["Marks"].idxmax()
```

```python
df.loc[df["Marks"].idxmax()]
```

```python
df.loc[df["Marks"].idxmin()]
```

```python
df["City"].value_counts()
```

```python
(df["Marks"] >= 40).sum()
```

```python
df.loc[df["City"] == "Mumbai", "Marks"].mean()
```

| Task                | Code                                          |
| ------------------- | --------------------------------------------- |
| Average             | `df["Marks"].mean()`                          |
| Median              | `df["Marks"].median()`                        |
| Mode                | `df["Marks"].mode()`                          |
| Total               | `df["Marks"].sum()`                           |
| Minimum             | `df["Marks"].min()`                           |
| Maximum             | `df["Marks"].max()`                           |
| Count               | `df["Marks"].count()`                         |
| Standard deviation  | `df["Marks"].std()`                           |
| Variance            | `df["Marks"].var()`                           |
| Statistical summary | `df["Marks"].describe()`                      |
| Maximum row         | `df.loc[df["Marks"].idxmax()]`                |
| Minimum row         | `df.loc[df["Marks"].idxmin()]`                |
| Multiple statistics | `df["Marks"].agg(["mean","min","max"])`       |
| Category counts     | `df["City"].value_counts()`                   |
| Count passed        | `(df["Marks"] >= 40).sum()`                   |
| Filter + analysis   | `df.loc[df["City"]=="Mumbai","Marks"].mean()` |
