```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})

df
```

```python
df.sort_values("Marks")
```

```python
df.sort_values("Marks", ascending=False)
```

```python
df.sort_values("Name")
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95]
})
```

```python
df.sort_values(["Age", "Marks"])
```

```python
df.sort_values(
    ["Age", "Marks"],
    ascending=[True, False]
)
```

```python
df = pd.DataFrame(
    {"Marks": [85, 92, 76]},
    index=["C", "A", "B"]
)

df
```

```python
df.sort_index()
```

```python
df.nlargest(3, "Marks")
```

```python
df.nsmallest(1, "Marks")
```

| Task             | Code                                                      |
| ---------------- | --------------------------------------------------------- |
| Sort column      | `df.sort_values("Marks")`                                 |
| Descending       | `df.sort_values("Marks", ascending=False)`                |
| Multiple columns | `df.sort_values(["Age", "Marks"])`                        |
| Different orders | `df.sort_values(["Age","Marks"], ascending=[True,False])` |
| Sort index       | `df.sort_index()`                                         |
| Top 3            | `df.nlargest(3, "Marks")`                                 |
| Bottom 3         | `df.nsmallest(3, "Marks")`                                |
