import pandas as pd

# ==========================================
# 1. Create DataFrame
# ==========================================

df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})

print("Original DataFrame:")
print(df)


# ==========================================
# 2. Save DataFrame as CSV
# ==========================================

df.to_csv("students.csv", index=False)


# ==========================================
# 3. Read CSV File
# ==========================================

df2 = pd.read_csv("students.csv")

print("\nData read from CSV:")
print(df2)


# ==========================================
# 4. Basic Information
# ==========================================

print("\nFirst 5 rows:")
print(df2.head())

print("\nShape:")
print(df2.shape)

print("\nColumns:")
print(df2.columns)

print("\nData Types:")
print(df2.dtypes)

print("\nStatistical Summary:")
print(df2.describe())


# ==========================================
# 5. Read Specific Columns
# ==========================================

df3 = pd.read_csv(
    "students.csv",
    usecols=["Name", "Marks"]
)

print("\nSelected Columns:")
print(df3)


# ==========================================
# 6. Modify Data
# ==========================================

df2["Passed"] = df2["Marks"] >= 40

print("\nData after adding Passed column:")
print(df2)


# ==========================================
# 7. Save Modified Data
# ==========================================

df2.to_csv("students_updated.csv", index=False)


# ==========================================
# 8. Save Selected Columns
# ==========================================

df2[["Name", "Marks"]].to_csv(
    "marks.csv",
    index=False
)


# ==========================================
# 9. Different Separator
# ==========================================

df.to_csv(
    "students_semicolon.csv",
    sep=";",
    index=False
)

df4 = pd.read_csv(
    "students_semicolon.csv",
    sep=";"
)

print("\nData with semicolon separator:")
print(df4)


# ==========================================
# 10. Missing Values
# ==========================================

df_missing = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit"],
    "Marks": [85, "NA", 76]
})

df_missing.to_csv(
    "students_missing.csv",
    index=False
)

df5 = pd.read_csv(
    "students_missing.csv",
    na_values=["NA"]
)

print("\nData with missing value:")
print(df5)

print("\nMissing Values:")
print(df5.isna().sum())


# ==========================================
# 11. Excel File
# ==========================================

df.to_excel(
    "students.xlsx",
    sheet_name="Students",
    index=False
)

excel_df = pd.read_excel(
    "students.xlsx",
    sheet_name="Students"
)

print("\nData read from Excel:")
print(excel_df)


# ==========================================
# 12. Analysis and Save Result
# ==========================================

average_marks = df["Marks"].mean()

print("\nAverage Marks:", average_marks)

passed = df[df["Marks"] >= 40]

print("\nPassed Students:")
print(passed)

passed.to_csv(
    "passed_students.csv",
    index=False
)

print("\nFiles created successfully!")

pd.read_csv()       → Read CSV
df.to_csv()         → Save CSV

pd.read_excel()     → Read Excel
df.to_excel()       → Save Excel

usecols             → Read selected columns
sep                 → Change separator
na_values           → Identify missing values
index=False         → Don't save DataFrame index
