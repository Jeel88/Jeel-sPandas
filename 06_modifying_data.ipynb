```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})
```

```python
df.loc[0,"Marks"]=90
```

```python
df.iloc[0, 2] = 90
```

```python
df["Marks"] = df["Marks"] + 5
```

```python
df["Passed"] = df["Marks"] >= 40
```

```python
df
```

```python
df = pd.DataFrame({
    "Name": ["Jeel", "Rahul", "Amit", "Priya", "Neha"],
    "Age": [20, 21, 19, 20, 22],
    "Marks": [85, 92, 76, 88, 95],
    "City": ["Mumbai", "Pune", "Surat", "Mumbai", "Delhi"]
})
df
```

**df.loc[condition, "column"] = new_value**

```python
df.loc[df["Marks"] < 80, "Marks"] += 5
```

```python
df
```

```python
df["City"] = df["City"].replace("Mumbai", "Bombay")
```

```python
df["City"] = df["City"].replace({
    "Mumbai": "Bombay",
    "Pune": "Poona"
})
df
```

```python
df = df.rename(columns={
    "Name": "Student_Name",
    "Age": "Student_Age"
})
df
```

```python
df = df.rename(index={
    0: "A",
    1: "B",
    2: "C",
    3: "D",
    4: "E"
})
```

| Task                             | Code                                                              |
| -------------------------------- | ----------------------------------------------------------------- |
| Change single value              | `df.loc[0, "Marks"] = 90`                                         |
| Change by position               | `df.iloc[0, 2] = 90`                                              |
| Modify entire column             | `df["Marks"] = df["Marks"] + 5`                                   |
| Add new column                   | `df["Passed"] = df["Marks"] >= 40`                                |
| Conditional update               | `df.loc[df["Marks"] < 40, "Marks"] = 40`                          |
| Conditional update + calculation | `df.loc[df["Marks"] < 80, "Marks"] += 5`                          |
| Replace value                    | `df["City"] = df["City"].replace("Mumbai", "Bombay")`             |
| Multiple replacements            | `df["City"].replace({"Mumbai":"Bombay", "Pune":"Poona"})`         |
| Rename column                    | `df.rename(columns={"Marks":"Score"})`                            |
| Rename multiple columns          | `df.rename(columns={"Name":"Student_Name", "Age":"Student_Age"})` |
| Rename index                     | `df.rename(index={0:"A", 1:"B"})`                                 |
| Add row                          | `df.loc["F"] = ["Karan", 21, 82, "Ahmedabad"]`                    |
| Delete row                       | `df.drop("F")`                                                    |
| Delete column                    | `df.drop(columns="Age")`                                          |
| Add column at position           | `df.insert(1, "Country", "India")`                                |
