```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    "Name": ["Jeel", "Rahul"],
    "Marks": [85, 92]
})

df2 = pd.DataFrame({
    "NAme": ["Amit", "Priya"],
    "Marks": [76, 88]
})

df1
```

```python
df2
```

```python
result = pd.concat([df1, df2])
result
```

```python
result = pd.concat([df1, df2], ignore_index=True)

result
```

```python
df1 = pd.DataFrame({
    "Name": ["Jeel", "Rahul"],
    "Marks": [85, 92]
})

df2 = pd.DataFrame({
    "City": ["Mumbai", "Pune"],
    "Age": [20, 21]
})

pd.concat([df1, df2], axis=1)
```

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Jeel", "Rahul", "Amit"]
})
marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 92, 76]
})
```

```python
pd.merge(students,marks,on="ID")
```

```python
pd.merge(students, marks, on="ID", how="inner")
```

```python
pd.merge(students, marks, on="ID", how="outer")
```

```python
pd.merge(students, marks, on="ID", how="left")
```

```python
pd.merge(students, marks, on="ID", how="right")
```

```python
df1 = pd.DataFrame({
    "Name": ["Jeel", "Rahul"],
    "Marks": [85, 92]
}, index=[1, 2])

df2 = pd.DataFrame({
    "City": ["Mumbai", "Pune"]
}, index=[1, 2])
```

```python
df1.join(df2)
df1
```

# Rows
pd.concat([df1, df2])

# Rows + fresh index
pd.concat([df1, df2], ignore_index=True)

# Columns
pd.concat([df1, df2], axis=1)

# Merge using common column
pd.merge(df1, df2, on="ID")

# Inner
pd.merge(df1, df2, on="ID", how="inner")

# Left
pd.merge(df1, df2, on="ID", how="left")

# Right
pd.merge(df1, df2, on="ID", how="right")

# Outer
pd.merge(df1, df2, on="ID", how="outer")

# Join using index
df1.join(df2)
